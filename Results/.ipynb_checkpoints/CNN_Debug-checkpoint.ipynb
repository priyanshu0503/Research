## Spectrum Reader

Reads one `.txt` spectrum file and returns Raman shift values and dark-subtracted intensities. Skips uncalibrated pixels (empty wavelength column).

```python
def read_spectrum(filepath):
    raman_shifts = []
    intensities  = []

    with open(filepath, 'r', encoding='latin-1') as f:
        lines = f.readlines()

    data_started = False
    for line in lines:
        line = line.strip()
        if line.startswith("Pixel;"):
            data_started = True
            continue
        if not data_started:
            continue
        parts = line.split(';')
        if len(parts) < 8:
            continue
        wl_str = parts[1].strip()
        if not wl_str or wl_str == '   ':
            continue
        try:
            raman_shift = float(parts[3])
            dark_sub    = float(parts[7])
            raman_shifts.append(raman_shift)
            intensities.append(dark_sub)
        except ValueError:
            continue

    return raman_shifts, intensities
```

## Load All Spectra

Recursively loads all `.txt` spectrum files for each cultivar. Each cultivar should have 324 spectra.

```python
all_data = {}

for folder_name, meta in CULTIVARS.items():
    folder_path = DATASET_DIR / folder_name
    if not folder_path.exists():
        print(f"Not found: {folder_name}")
        continue
    files   = list(folder_path.rglob("*.txt"))
    spectra = []
    for f in files:
        rs, intens = read_spectrum(f)
        if len(rs) > 100:
            spectra.append((rs, intens))
    all_data[folder_name] = spectra
    print(f"{meta['label']}: {len(spectra)} spectra loaded")
```

## Preprocessing Functions

Defines all preprocessing functions used across experiments: Savitzky-Golay smoothing, SNV normalisation, MSC correction, and first and second derivatives. SG + SNV was identified as the best combination across all models in the full-label regime.

```python
from scipy.signal import savgol_filter

def savitzky_golay_smooth(X, window=11, poly=3):
    """Smooth each spectrum to reduce noise"""
    return savgol_filter(X, window_length=window,
                         polyorder=poly, axis=1)

def snv(X):
    """Standard Normal Variate — removes intensity scaling differences"""
    mean = X.mean(axis=1, keepdims=True)
    std  = X.std(axis=1, keepdims=True)
    return (X - mean) / std

def msc(X):
    """Multiplicative Scatter Correction"""
    mean_spectrum = X.mean(axis=0)
    X_msc = np.zeros_like(X)
    for i in range(X.shape[0]):
        coef = np.polyfit(mean_spectrum, X[i], 1)
        X_msc[i] = (X[i] - coef[1]) / coef[0]
    return X_msc

def first_derivative(X, window=11, poly=3):
    """First derivative — removes baseline offset"""
    return savgol_filter(X, window_length=window,
                         polyorder=poly, deriv=1, axis=1)

def second_derivative(X, window=11, poly=3):
    """Second derivative — stronger baseline removal"""
    return savgol_filter(X, window_length=window,
                         polyorder=poly, deriv=2, axis=1)

preprocessing_combos = {
    "SG + SNV":           lambda X: snv(savitzky_golay_smooth(X)),
    "SG + MSC":           lambda X: msc(savitzky_golay_smooth(X)),
    "SG + 1st Deriv":     lambda X: first_derivative(savitzky_golay_smooth(X)),
    "SG + 2nd Deriv":     lambda X: second_derivative(savitzky_golay_smooth(X)),
    "SG + SNV + 1st Der": lambda X: first_derivative(snv(savitzky_golay_smooth(X))),
    "SG + SNV + 2nd Der": lambda X: second_derivative(snv(savitzky_golay_smooth(X))),
    "SG + MSC + 1st Der": lambda X: first_derivative(msc(savitzky_golay_smooth(X))),
    "SG + MSC + 2nd Der": lambda X: second_derivative(msc(savitzky_golay_smooth(X))),
}

print(f"Preprocessing functions defined: {len(preprocessing_combos)} combinations ready")
```

## Build X and y Matrices

Stacks all spectra into a matrix X of shape (6480, 1530) and a label vector y of shape (6480,). Each row in X corresponds to one spectrum and each value in y is the HPLC cysteine concentration for that cultivar.

```python
X = []
y = []

for folder_name, meta in CULTIVARS.items():
    spectra = all_data.get(folder_name, [])
    for rs, intens in spectra:
        X.append(intens)
        y.append(meta['cys'])

X = np.array(X)
y = np.array(y)

print(f"X shape: {X.shape}")
print(f"y shape: {y.shape}")
print(f"Cysteine range: {y.min():.4f} to {y.max():.4f}")
```

## Apply Preprocessing

Applies SG smoothing followed by SNV normalisation to the full dataset. This is the best preprocessing combination identified across PLSR, SVR, RF and CNN experiments.

```python
X_pre = snv(savitzky_golay_smooth(X))

print(f"Preprocessed X shape: {X_pre.shape}")
print(f"Mean of first spectrum after SNV: {X_pre[0].mean():.4f}  (should be ~0)")
print(f"Std  of first spectrum after SNV: {X_pre[0].std():.4f}   (should be ~1)")
```

## CNN Architecture — 1D Convolutional Neural Network

Four convolutional blocks each containing a Conv1d layer, batch normalisation, ReLU activation and max pooling. The spectrum is treated as a 1D signal and the network learns local peak features automatically. Batch normalisation handles intensity scaling internally, reducing dependence on external preprocessing.

```python
class CNN1D(nn.Module):
    def __init__(self, input_size=1530):
        super(CNN1D, self).__init__()

        self.conv_blocks = nn.Sequential(
            nn.Conv1d(1, 16, kernel_size=5, padding=2),
            nn.BatchNorm1d(16),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(16, 32, kernel_size=5, padding=2),
            nn.BatchNorm1d(32),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(32, 64, kernel_size=5, padding=2),
            nn.BatchNorm1d(64),
            nn.ReLU(),
            nn.MaxPool1d(2),

            nn.Conv1d(64, 128, kernel_size=5, padding=2),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.MaxPool1d(2),
        )

        self.flat_size = 128 * (input_size // 16)

        self.fc_layers = nn.Sequential(
            nn.Linear(self.flat_size, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, 1)
        )

    def forward(self, x):
        x = x.unsqueeze(1)
        x = self.conv_blocks(x)
        x = x.view(x.size(0), -1)
        x = self.fc_layers(x)
        return x.squeeze(1)

# Test
model_test = CNN1D(input_size=1530)
dummy = torch.randn(4, 1530)
out   = model_test(dummy)
print(f"Total parameters: {sum(p.numel() for p in model_test.parameters()):,}")
print(f"Output shape: {out.shape}  ← should be (4,)")
```

## Fixed Training Function — No Data Leakage

This is the corrected version of the original `train_cnn` function. The key change is that no validation set is passed during LOCO training. The original version used the test cultivar as the validation set and saved the checkpoint with the lowest validation loss — meaning the model was tuned to the test cultivar it was supposed to have never seen. This fix trains for a fixed number of epochs with no checkpoint selection, ensuring the test cultivar has zero influence on training.

```python
def train_cnn_fixed(model, X_train, y_train,
                    epochs=100, batch_size=32, lr=1e-4):
    """
    FIXED version — no validation set, no checkpoint selection.
    Test cultivar is strictly isolated from all training decisions.
    Trains for a fixed epoch budget only.
    """
    X_tr = torch.FloatTensor(X_train).to(device)
    y_tr = torch.FloatTensor(y_train).to(device)

    dataset  = TensorDataset(X_tr, y_tr)
    loader   = DataLoader(dataset, batch_size=batch_size, shuffle=True)

    criterion = nn.SmoothL1Loss(beta=0.02)
    optimizer = optim.AdamW(model.parameters(),
                            lr=lr, weight_decay=1e-4)
    scheduler = optim.lr_scheduler.OneCycleLR(
                    optimizer, max_lr=lr,
                    steps_per_epoch=len(loader),
                    epochs=epochs)

    model.to(device)

    for epoch in range(epochs):
        model.train()
        epoch_loss = 0
        for X_batch, y_batch in loader:
            optimizer.zero_grad()
            y_pred = model(X_batch)
            loss   = criterion(y_pred, y_batch)
            loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            scheduler.step()
            epoch_loss += loss.item()

        if (epoch + 1) % 10 == 0:
            avg = epoch_loss / len(loader)
            print(f"  Epoch {epoch+1:3d}/{epochs} — "
                  f"Train Loss: {avg:.5f}")

    return model

print("Fixed train_cnn_fixed defined!")
print("No validation set — no test data leakage possible!")
```

## Corrected LOCO Cross Validation

Runs Leave-One-Cultivar-Out cross validation with the leakage bug fixed. In each fold the held-out test cultivar is used exclusively for final evaluation, it never appears in the training data or influences any training decision. The model trains on 19 cultivars for a fixed 100 epochs and then predicts on the 20th.

```python
all_y_true_fixed = []
all_y_pred_fixed = []
loco_rmse_fixed  = []

for fold_idx, held_out in enumerate(cultivar_names):
    print(f"Fold {fold_idx+1}/20 — "
          f"Holding out: {CULTIVARS[held_out]['label']}")

    X_tr_l, y_tr_l = [], []
    X_te_l, y_te_l = [], []

    for folder_name, meta in CULTIVARS.items():
        spectra = all_data.get(folder_name, [])
        for rs, intens in spectra:
            arr     = np.array([intens])
            arr_pre = snv(savitzky_golay_smooth(arr))[0]
            if folder_name == held_out:
                X_te_l.append(arr_pre)
                y_te_l.append(meta['cys'])
            else:
                X_tr_l.append(arr_pre)
                y_tr_l.append(meta['cys'])

    X_tr_l = np.array(X_tr_l)
    y_tr_l = np.array(y_tr_l)
    X_te_l = np.array(X_te_l)
    y_te_l = np.array(y_te_l)

    # Train on 19 cultivars — test cultivar never touched
    torch.manual_seed(42)
    model_fixed = CNN1D(input_size=1530).to(device)
    model_fixed = train_cnn_fixed(
        model=model_fixed,
        X_train=X_tr_l,
        y_train=y_tr_l,
        epochs=100,
        batch_size=32,
        lr=1e-4
    )

    # Predict on held-out test cultivar
    model_fixed.eval()
    with torch.no_grad():
        y_pred_l = model_fixed(
            torch.FloatTensor(X_te_l).to(device)
        ).cpu().numpy()

    fold_rmse = np.sqrt(mean_squared_error(y_te_l, y_pred_l))
    loco_rmse_fixed.append(fold_rmse)

    print(f"  RMSE: {fold_rmse:.4f}  "
          f"True: {y_te_l[0]:.4f}  "
          f"Predicted mean: {np.mean(y_pred_l):.4f}\n")

    all_y_true_fixed.extend(y_te_l)
    all_y_pred_fixed.extend(y_pred_l)

all_y_true_fixed = np.array(all_y_true_fixed)
all_y_pred_fixed = np.array(all_y_pred_fixed)

rmse_fixed = np.sqrt(mean_squared_error(all_y_true_fixed, all_y_pred_fixed))
r2_fixed   = r2_score(all_y_true_fixed, all_y_pred_fixed)
rpd_fixed  = all_y_true_fixed.std() / rmse_fixed

print(f"\n=== CORRECTED LOCO Results ===")
print(f"Global RMSE: {rmse_fixed:.4f}")
print(f"Global R²:   {r2_fixed:.4f}")
print(f"Global RPD:  {rpd_fixed:.4f}")
print(f"\nPrevious (leaked) LOCO R²:  0.6128")
print(f"Corrected (clean) LOCO R²:  {r2_fixed:.4f}")
```

## Corrected LOCO Results — Scatter Plot and Comparison

Visualises the corrected LOCO predictions coloured by cultivar. Prints a side by side comparison of the original leaked results against the clean fixed results to quantify the impact of the bug.

```python
colors = plt.cm.tab20(np.linspace(0, 1, 20))
fig, ax = plt.subplots(figsize=(8, 7))

unique_cys = sorted(set(all_y_true_fixed))
for i, val in enumerate(unique_cys):
    mask  = all_y_true_fixed == val
    label = [meta['label'] for meta in CULTIVARS.values()
             if abs(meta['cys'] - val) < 0.0001]
    label = label[0] if label else str(round(val, 4))
    ax.scatter(all_y_true_fixed[mask], all_y_pred_fixed[mask],
               alpha=0.4, s=10, color=colors[i], label=label)

min_val = min(all_y_true_fixed.min(), all_y_pred_fixed.min()) - 0.005
max_val = max(all_y_true_fixed.max(), all_y_pred_fixed.max()) + 0.005
ax.plot([min_val, max_val], [min_val, max_val],
        'r--', linewidth=1.5, label='Perfect prediction')
ax.set_xlabel("True Cysteine (g/100g)")
ax.set_ylabel("Predicted Cysteine (g/100g)")
ax.set_title(f"CNN LOCO — Corrected (No Data Leakage)\n"
             f"RMSE={rmse_fixed:.4f}  R²={r2_fixed:.4f}  RPD={rpd_fixed:.4f}")
ax.legend(fontsize=6, ncol=2, loc='upper left')
plt.tight_layout()
plt.savefig("CNN_LOCO_fixed.png", dpi=150, bbox_inches='tight')
plt.show()

print("\n" + "="*55)
print("   BUGGY vs FIXED — LOCO COMPARISON")
print("="*55)
print(f"{'Metric':<20} {'Buggy (leaked)':>16} {'Fixed (clean)':>14}")
print("-"*55)
print(f"{'RMSE':<20} {'0.0102':>16} {rmse_fixed:>14.4f}")
print(f"{'R²':<20} {'0.6128':>16} {r2_fixed:>14.4f}")
print(f"{'RPD':<20} {'1.6071':>16} {rpd_fixed:>14.4f}")
print("="*55)
if r2_fixed < 0.6128:
    print(f"\nR² dropped by {0.6128 - r2_fixed:.4f} — leakage was inflating results")
elif r2_fixed >= 0.6128:
    print(f"\nR² unchanged or improved — leakage had minimal effect")
```
